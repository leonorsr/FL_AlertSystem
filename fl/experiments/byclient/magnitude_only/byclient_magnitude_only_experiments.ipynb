Use this notebook to inspect the outputs generated by `run_baseline_suite.py`.

```python
from pathlib import Path
import pandas as pd

results_root = Path('results')
run_dirs = sorted([path for path in results_root.glob('run_*') if path.is_dir()])
run_dirs[-3:]
```

```python
latest_run = run_dirs[-1]
metrics_global = pd.read_csv(latest_run / 'metrics_global.csv')
metrics_by_dataset = pd.read_csv(latest_run / 'metrics_by_dataset.csv')
model_selection_dev = pd.read_csv(latest_run / 'model_selection_dev.csv')
chosen_models = pd.read_csv(latest_run / 'chosen_models.csv')
split_summary = pd.read_csv(latest_run / 'split_summary.csv')

display(split_summary)
display(chosen_models)
display(metrics_global.sort_values('f1', ascending=False))
```

## Final Comparison Table

This table keeps only the global `test` results and highlights the main metrics for each model.

```python
final_comparison = (
    metrics_global
    [[
        'model',
        'selected_candidate',
        'accuracy',
        'balanced_accuracy',
        'specificity',
        'precision',
        'recall',
        'f1',
        'roc_auc',
        'pr_auc',
        'far',
        'miss_rate',
        'tn',
        'fp',
        'fn',
        'tp',
    ]]
    .sort_values(
        ['pr_auc', 'miss_rate', 'far', 'balanced_accuracy', 'f1'],
        ascending=[False, True, True, False, False],
    )
    .reset_index(drop=True)
)

final_comparison
```

```python
styled_final_comparison = (
    final_comparison.style
    .format({
        'accuracy': '{:.4f}',
        'balanced_accuracy': '{:.4f}',
        'specificity': '{:.4f}',
        'precision': '{:.4f}',
        'recall': '{:.4f}',
        'f1': '{:.4f}',
        'roc_auc': '{:.4f}',
        'pr_auc': '{:.4f}',
        'far': '{:.4f}',
        'miss_rate': '{:.4f}',
    })
    .highlight_max(subset=['accuracy', 'balanced_accuracy', 'specificity', 'precision', 'recall', 'f1', 'roc_auc', 'pr_auc'], color='#d8f3dc')
    .highlight_min(subset=['far', 'miss_rate'], color='#fde2e4')
)

styled_final_comparison
```

```python
final_comparison.to_csv(latest_run / 'final_model_comparison.csv', index=False)
print(f'Saved comparison table to: {latest_run / "final_model_comparison.csv"}')
```

## Interpretation of Results

This section summarizes the global ranking and the main differences observed across datasets using the metrics that are most relevant for fall detection.

```python
from IPython.display import Markdown, display

ranking_columns = ['pr_auc', 'miss_rate', 'far', 'balanced_accuracy', 'f1']
ranking_ascending = [False, True, True, False, False]

global_rank = metrics_global.sort_values(ranking_columns, ascending=ranking_ascending).reset_index(drop=True)
best_global = global_rank.iloc[0]
runner_up = global_rank.iloc[1]

best_per_dataset = (
    metrics_by_dataset
    .sort_values(['dataset', *ranking_columns], ascending=[True, *ranking_ascending])
    .groupby('dataset', as_index=False)
    .first()
)

hardest_dataset = best_per_dataset.sort_values(['pr_auc', 'miss_rate'], ascending=[True, False]).iloc[0]
easiest_dataset = best_per_dataset.sort_values(['pr_auc', 'miss_rate'], ascending=[False, True]).iloc[0]
lowest_far = metrics_global.sort_values('far', ascending=True).iloc[0]
lowest_miss_rate = metrics_global.sort_values('miss_rate', ascending=True).iloc[0]
highest_specificity = metrics_global.sort_values('specificity', ascending=False).iloc[0]

summary_lines = [
    '### Key Takeaways',
    '',
    (
        f"- **Best overall model under the fall-detection ranking:** `{best_global['model']}` ({best_global['selected_candidate']}) comes first because it combines the highest PR-AUC (**{best_global['pr_auc']:.4f}**) with a very low miss rate (**{best_global['miss_rate']:.4f}**) and a low FAR (**{best_global['far']:.4f}**). In this context, that makes it the strongest option for detecting falls while keeping false alarms controlled."
    ),
    (
        f"- **Closest competitor:** `{runner_up['model']}` is also very strong, with PR-AUC = **{runner_up['pr_auc']:.4f}** and FAR = **{runner_up['far']:.4f}**. However, its miss rate (**{runner_up['miss_rate']:.4f}**) is higher than the top-ranked model, which is an important limitation when missing a fall is more costly than raising an alert."
    ),
    (
        f"- **Dataset difficulty:** `{easiest_dataset['dataset']}` is the easiest dataset in the current setup (best PR-AUC = **{easiest_dataset['pr_auc']:.4f}**, miss rate = **{easiest_dataset['miss_rate']:.4f}**), while `{hardest_dataset['dataset']}` is clearly the most challenging (best PR-AUC = **{hardest_dataset['pr_auc']:.4f}**, miss rate = **{hardest_dataset['miss_rate']:.4f}**)."
    ),
    (
        f"- **Error trade-offs:** `{lowest_miss_rate['model']}` achieves the lowest miss rate (**{lowest_miss_rate['miss_rate']:.4f}**), which is desirable in a safety-critical fall detection setting. `{lowest_far['model']}` has the lowest FAR (**{lowest_far['far']:.4f}**), and `{highest_specificity['model']}` reaches the highest specificity (**{highest_specificity['specificity']:.4f}**), so these models are preferable when reducing false positives is the main concern."
    ),
    (
        '- **Overall interpretation:** the nonlinear models remain the strongest baselines, with the neural network and XGBoost clearly outperforming logistic regression and kNN on the combined test set. The main performance drop appears on UpFall, where all models lose substantial precision-recall quality compared with KFall and SisFall.'
    ),
]

display(Markdown('\n'.join(summary_lines)))
```
